```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
events_df = pd.read_csv('../data/raw/events.csv')
events_df.head()
```

```python
funnel_counts = events_df['event'].value_counts()

print(funnel_counts)
```

```python
funnel_order = [
    'signup',
    'activation',
    'feature_use',
    'purchase'
]

funnel_counts = funnel_counts[funnel_order]

#what is funnel ?
# A funnel is a visual representation of the customer journey, showing how users progress through different stages of interaction with a product or service. It helps identify where users drop off and where they convert, allowing businesses to optimize their processes and improve user experience.
```

```python
print(funnel_counts)
```

```python
conversion_rates = []

for i in range(len(funnel_counts)-1):

    current_stage = funnel_counts.iloc[i]
    next_stage = funnel_counts.iloc[i+1]

    conversion = (next_stage / current_stage) * 100

    conversion_rates.append(round(conversion, 2))

print(conversion_rates)

#explain in detail what is conversion rate and how to calculate it ?

# Conversion rate is a key performance metric that measures the percentage of users who take a desired action (such as signing up, activating, using a feature, or making a purchase) compared to the total number of users at the previous stage of the funnel. It helps businesses understand how effectively they are converting potential customers into actual customers. 
```

```python
for i in range(len(funnel_order)-1):

    print(
        f"{funnel_order[i]} → {funnel_order[i+1]}: "
        f"{conversion_rates[i]}%"
    )
    #there are 100% so how this distribition?
    # 
```

```python
drop_off_rates = []

for rate in conversion_rates:

    drop_off = 100 - rate
    drop_off_rates.append(round(drop_off, 2))

print(drop_off_rates)
```

```python
plt.figure(figsize=(10,6))

plt.bar(
    funnel_counts.index,
    funnel_counts.values
)

plt.title('Product Funnel Analysis')
plt.xlabel('Funnel Stage')
plt.ylabel('Number of Users')

plt.show()

#The funnel chart visualizes user progression across key product stages and helps identify where the largest behavioral drop-offs occur. 
```

```python
largest_drop = max(drop_off_rates)

index = drop_off_rates.index(largest_drop)

print(
    f"Largest drop-off occurs between "
    f"{funnel_order[index]} and "
    f"{funnel_order[index+1]}"
)
```

```python
#Stage labels
stages = [
    'Signup → Activation',
    'Activation → Feature',
    'Feature → Purchase'
]


# Create figure
plt.figure(figsize=(10,6))

# Plot conversion line
plt.plot(
    stages,
    conversion_rates,
    marker='o',
    label='Conversion Rate'
)

# Plot drop-off line
plt.plot(
    stages,
    drop_off_rates,
    marker='o',
    label='Drop-Off Rate'
)

# Titles and labels
plt.title('Conversion vs Drop-Off Rates')
plt.xlabel('Funnel Stages')
plt.ylabel('Percentage')

# Show legend
plt.legend()

# Show grid
plt.grid(True)

# Show chart
plt.show()
```
