```python
import pandas as pd
import numpy as np
from faker import Faker # For generating fake data
from datetime import datetime, timedelta
import random # For random date generation
```

```python
fake = Faker() # Initialize Faker to generate fake data
fake.name() # Example of generating a fake name using Faker
```

```python
NUM_USERS = 5000
```

```python
users = []
events = []

# Generate user data
```

```python
for user_id in range(1, NUM_USERS + 1):

    signup_date = fake.date_time_between(
        start_date='-90d',
        end_date='now'
    )

    variant = random.choice(['A', 'B']) # Randomly assign users to variant A or B

    users.append({
        'user_id': user_id,
        'signup_date': signup_date,
        'variant': variant
    })

# Generate event data
```

```python
users_df = pd.DataFrame(users)
users_df.head()
```

```python
for _, row in users_df.iterrows(): # Iterate over each user to generate their event data

    user_id = row['user_id']
    signup_date = row['signup_date']
    variant = row['variant']

    # Every user signs up
    events.append({
        'user_id': user_id,
        'event': 'signup',
        'timestamp': signup_date,
        'variant': variant
    })

    # Activation probability
    activated = np.random.rand() < 0.65 # 65% chance of activation

    if activated:

        activation_time = signup_date + timedelta(
            hours=random.randint(1, 48)
        ) # Activation happens within 1 to 48 hours after signup

        events.append({
            'user_id': user_id,
            'event': 'activation',
            'timestamp': activation_time,
            'variant': variant
        })

        # Feature usage probability
        used_feature = np.random.rand() < 0.50 # 50% chance of using the feature after activation

        if used_feature: 

            feature_time = activation_time + timedelta(
                days=random.randint(1, 7)
            )

            events.append({
                'user_id': user_id,
                'event': 'feature_use',
                'timestamp': feature_time,
                'variant': variant
            })

            # Purchase probability
            purchased = np.random.rand() < 0.30 # 30% chance of making a purchase after using the feature

            if purchased:

                purchase_time = feature_time + timedelta(
                    days=random.randint(1, 10)
                )

                events.append({
                    'user_id': user_id,
                    'event': 'purchase',
                    'timestamp': purchase_time,
                    'variant': variant
                })

# Convert events to DataFrame
```

```python
events_df = pd.DataFrame(events)
events_df.head(20)

```

```python
users_df.to_csv('../data/raw/users.csv', index=False)

events_df.to_csv('../data/raw/events.csv', index=False)
```
